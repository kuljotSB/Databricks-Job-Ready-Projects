### Set up your Environment

```python
from databricks.sdk import WorkspaceClient

# Get Databricks runtime authentication
w = WorkspaceClient()

headers = w.config.authenticate()
token = headers["Authorization"].replace("Bearer ", "")
workspace_host = w.config.host.rstrip("/")
```

### Create an OpenAI Client

```python
from openai import OpenAI

client = OpenAI(
    api_key = token,
    base_url = f"{workspace_host}/serving-endpoints"
)
```

### Throwing a Request via the OpenAI API

```python
# OpenAI Request
completion = client.embeddings.create(
    model="databricks-gte-large-en",
    input="hi how are you?"
)

print("length \n")
print(len(completion.data[0].embedding))
print("\n embeddings: \n")
print(completion.data[0].embedding)
```
