### Load CSV (Volume) -> Bronze Tables

```python
def load_csv(path, table):
    (spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .csv(path)
        .write.mode("overwrite")
        .saveAsTable(table))

base = "/Volumes/carbonops/bronze/raw_files"

load_csv(f"{base}/carbon_targets.csv", "carbonops.bronze.carbon_targets")
load_csv(f"{base}/electricity_consumption.csv", "carbonops.bronze.electricity_consumption")
load_csv(f"{base}/electricity_corrections_incremental.csv", "carbonops.bronze.electricity_corrections_incremental")
load_csv(f"{base}/emission_factors.csv", "carbonops.bronze.emission_factors")
load_csv(f"{base}/facilities.csv", "carbonops.bronze.facilities")
load_csv(f"{base}/fuel_consumption.csv", "carbonops.bronze.fuel_consumption")
load_csv(f"{base}/production.csv", "carbonops.bronze.production")

print("Bronze tables created")
```
