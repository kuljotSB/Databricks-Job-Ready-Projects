## Notebook 1 - Load Data and Create the UC Structure

### Setup the Catalog and Volume

```python
# Create Bronze, Silver, and Gold Schemas
spark.sql("""
CREATE SCHEMA IF NOT EXISTS carbonops.bronze;
""")

spark.sql("""
CREATE SCHEMA IF NOT EXISTS carbonops.silver;
""")

spark.sql("""
CREATE SCHEMA IF NOT EXISTS carbonops.gold;
""")

# Create Volume for Raw Emissions Data
spark.sql("""
CREATE VOLUME IF NOT EXISTS carbonops.bronze.raw_files;
""")

print("CarbonOps catalog, schemas, and raw data volume created successfully!")
```

### Load Files from GitHub to UC Volume

```python
import requests

# Define the current catalog
catalog_name = "carbonops"

# Define the base path using the current catalog
volume_base = f"/Volumes/{catalog_name}/bronze/raw_files"

# List of files to download
files = ["carbon_targets.csv", 
         "electricity_consumption.csv", 
         "electricity_corrections_incremental.csv",
         "emission_factors.csv",
         "facilities.csv",
         "fuel_consumption.csv",
         "production.csv"]

# Download each file
for file in files:
    url = f"https://raw.githubusercontent.com/kuljotSB/Databricks-Job-Ready-Projects/refs/heads/main/Data-Pipelines/CarbonOps_Emissions_Dataset/{file}"
    response = requests.get(url)
    response.raise_for_status()

    # Write to Unity Catalog volume
    with open(f"{volume_base}/{file}", "wb") as f:
        f.write(response.content)
```
